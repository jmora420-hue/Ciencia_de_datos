```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Taxi_Trip_Pricing.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

aseguraos que los numeros son numeros


```python
columnas_numericas = [
    'Trip_Distance_km', 'Passenger_Count', 'Base_Fare',
    'Per_Km_Rate', 'Per_Minute_Rate', 'Trip_Duration_Minutes', 'Trip_Price'
]

for col in columnas_numericas:
    print(f"Valores únicos en {col}:", df[col].unique())

for col in columnas_numericas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.head()
```

```python
for col in columnas_numericas:
    df[col] = df[col].fillna(df[col].median())

df.head(20)
```

```python
df.info()
```

```python
columnas_categoricas = ['Time_of_Day', 'Day_of_Week', 'Traffic_Conditions', 'Weather']

for col in columnas_categoricas:
    print(f"Valores únicos en {col}:", df[col].unique())

import numpy as np
df[columnas_categoricas] = df[columnas_categoricas].replace(['?', ' ', ''], np.nan)

df.head(20)
```

```python
# Imputación de valores nulos utilizando la Moda para variables categóricas
for col in columnas_categoricas:
    moda = df[col].mode()[0]
    df[col] = df[col].fillna(moda)

df.head(20)
```
