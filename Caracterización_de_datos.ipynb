```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("/content/drive/MyDrive/Colab Notebooks/Ciencia de datos/Dataset_Maestro_SustainCrop.xlsx")
df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
print("Duplicados:", df.duplicated().sum())
```

```python
# Quitar espacios en los nombres de las columnas
df.columns=df.columns.str.strip()

# Quitar columnas vacías que sobran (las "Unnamed")
df=df.loc[:, ~df.columns.str.contains("Unnamed")]

# Quitar la columna TIPO DE CULTIVO porque está vacía
df=df.drop(columns=["TIPO DE CULTIVO"])

# Asegurar que el año sea número
df["AÑO"]=pd.to_numeric(df["AÑO"],errors="coerce")

# Quitar duplicados
df=df.drop_duplicates()

# Quitar filas sin resumen (lo necesitamos para el análisis de texto)
df=df.dropna(subset=["RESUMEN DE ARTÍCULO"])

# Estandarizar columnas de categorías
for col in ["DATASET","RELEVANCIA","CATEGORÍA SUSTAINCROP","ENFOQUE SOSTENIBILIDAD UE"]:
  df[col]=df[col].fillna("Sin dato")
  df[col]=df[col].astype(str).str.strip().str.title()

df_limpio=df.copy()
print("Filas y columnas después de limpiar:", df_limpio.shape)
df_limpio.isnull().sum()
```

```python
print(df_limpio["RELEVANCIA"].value_counts())
print()
print(df_limpio["DATASET"].value_counts())
print()
print(df_limpio["CATEGORÍA SUSTAINCROP"].value_counts())
print()
print(df_limpio["ENFOQUE SOSTENIBILIDAD UE"].value_counts().head(5))
print()
print(df_limpio["REVISTA"].value_counts().head(5))
print()
print(df_limpio["ENTE FINANCIADOR"].value_counts().head(5))
```

```python
for col in ["RELEVANCIA","DATASET","CATEGORÍA SUSTAINCROP"]:
  print("---", col, "---")
  print((df_limpio[col].value_counts(normalize=True)*100).round(1))
  print()
```

```python
# Sí/No -> 1/0
df_limpio["DATASET_NUM"]=df_limpio["DATASET"].map({"Sí":1,"No":0})

# Relevancia -> 3/2/1
df_limpio["RELEVANCIA_NUM"]=df_limpio["RELEVANCIA"].map({"Alta":3,"Media":2,"Baja":1})

# Largo del resumen (cantidad de caracteres)
df_limpio["LARGO_RESUMEN"]=df_limpio["RESUMEN DE ARTÍCULO"].astype(str).str.len()

df_limpio[["DATASET","DATASET_NUM","RELEVANCIA","RELEVANCIA_NUM","LARGO_RESUMEN"]].head()
```

```python
for col in ["RELEVANCIA_NUM","DATASET_NUM","LARGO_RESUMEN"]:
  print("---", col, "---")
  print("Promedio:", df_limpio[col].mean())
  print("Mediana:", df_limpio[col].median())
  print("Moda:", df_limpio[col].mode()[0])
  print("Mínimo:", df_limpio[col].min())
  print("Máximo:", df_limpio[col].max())
  print()
```

```python
for col in ["REVISTA","INSTITUCIONES_COLABORADORAS","DATASET","RELEVANCIA","CATEGORÍA SUSTAINCROP","ENFOQUE SOSTENIBILIDAD UE","ENTE FINANCIADOR"]:
  print(col, "->", df_limpio[col].mode()[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Palabras comunes que no aportan (inglés y español)
palabras_vacias=["which","their","these","those","about","where","other","between","through",
                 "sobre","entre","desde","donde","también","puede","pueden","study","results"]

# Solo palabras de 5 letras o más y las 15 más frecuentes
cv=CountVectorizer(max_features=15, token_pattern=r"\b\w{5,}\b", stop_words=palabras_vacias)
matriz=cv.fit_transform(df_limpio["RESUMEN DE ARTÍCULO"].astype(str))

# Lo pasamos a una tabla
df_vec=pd.DataFrame(matriz.toarray(), columns=cv.get_feature_names_out())
df_vec.head()
```

```python
frecuencias=df_vec.sum().sort_values(ascending=False)
frecuencias
```
