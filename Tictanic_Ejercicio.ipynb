```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/titanic.csv")
df.head()
```

```python
columnas_numericas = [
    'PassengerId', 'Survived', 'Pclass',
    'Age', 'SibSp', 'Parch', 'Fare'
]

for col in columnas_numericas:
    print(f"Valores únicos en {col}:", df[col].unique())

for col in columnas_numericas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.head(40)
```

```python
promediospGen = df.groupby('Sex')['Age'].mean().round()

def imputar_edad(fila):
    if pd.isnull(fila['Age']):
        return promediospGen[fila['Sex']]
    else:
        return fila['Age']

df['Age'] = df.apply(imputar_edad, axis=1)

df['Age'] = df['Age'].astype(int)

df.head(40)
```

```python
df['Fare'] = pd.to_numeric(df['Fare'], errors='coerce')
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
df['Fare'] = df['Fare'].round(2)
```

```python
#df['Ticket'] = df['Ticket'].astype(str).str.extract(r'(\d+)$')[0] #"Dejarlo de comentario a no ser que sea la primera limpieza"
df['Ticket'] = pd.to_numeric(df['Ticket'], errors='coerce')
df['Ticket'] = df['Ticket'].astype('Int64')
df.head(40)
```

```python
df['Cabin'] = df['Cabin'].astype(str).str.split(' ').str[0]

# 2. Al convertir a texto, los nulos se vuelven la palabra 'nan', los regresamos a nulos reales
df['Cabin'] = df['Cabin'].replace('nan', np.nan)

# 3. Aplicar Last Observation Carried Forward (LOCF) para rellenar vacíos con el último conocido
df['Cabin'] = df['Cabin'].ffill()

df.head(40)
```

```python
df.info()
```

```python
#Quitamos las filas que todavía tengan algún NaN
"""Como ya imputamos casi todo, los que quedan son tan poquitos que
es más práctico y seguro eliminarlos."""
df.dropna(inplace=True)

print(df.isnull().sum())
```
