```python
with open("/content/drive/MyDrive/Colab Notebooks/CaperucitaRoja.txt","r",encoding="utf-8") as archivo:
  cuento=archivo.read()
```

```python
cuento=cuento.lower()
cuento=cuento.replace(",","")
cuento=cuento.replace(";","")
cuento=cuento.replace(".","")
cuento=cuento.replace("?","")
cuento=cuento.replace("¿","")
cuento=cuento.replace("¡","")
cuento=cuento.replace("!","")
cuento=cuento.replace("—","")
cuento=cuento.replace(":","")
cuento=cuento.replace('“',"")
palabras=cuento.split()
```

```python
import pandas as pd
import numpy as np

palabras_array=np.array(palabras)
longitudes=np.array([len(palabra) for palabra in palabras_array])

df=pd.DataFrame(
    {
        "palabra":palabras,
        "longitud":longitudes
    }
)
```

```python
stopwords = [
    "el","que","en","de","del","y","la","a","los","lo","al","muy","un","te","es","se","no","si","le","su",'mi',"por","o","son","eso","con","para","una","más","soy","así","las"
    ,"este","él","tan","con","esa","qué","toc","mi","pues","quien","muy","allá","porque"
]
```

```python
df_limpio=df[~df["palabra"].isin(stopwords)]
frecuencias=df_limpio["palabra"].value_counts()
frecuencias.head(20)
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from PIL import Image

texto_limpio=" ".join(df_limpio["palabra"])
mascara = np.array(Image.open("/content/drive/MyDrive/Colab Notebooks/caperusa.png"))

nube=WordCloud(
  width=1200,
  height=600,
  background_color="white",
  mask=mascara,
  colormap="Oranges_r"
).generate(texto_limpio)

plt.figure(figsize=(15,7))
plt.imshow(nube)
plt.axis("off")
plt.show()
```
